### (7.1) Time Series Analysis: Introduction

Time series analysis is a specialized branch of statistics and data science focused on analyzing data points collected over a period of time. The primary goal is to understand patterns, trends, and seasonality within the data to improve forecasts, identify anomalies, and gain insights into the underlying processes generating the observations. Unlike cross-sectional data, time series data inherently has a temporal order, meaning the sequence of observations is crucial. This sequential dependency often leads to techniques that account for past values influencing future ones, such as autocorrelation and lag. Applications span diverse fields, from predicting stock prices and forecasting demand to understanding consumer behavior and disease outbreaks.

### Moving Averages and Business Applications

A **moving average (MA)** is a simple and widely used technique in time series analysis to smooth out short-term fluctuations and highlight longer-term trends or cycles. It calculates the average of a specific number of consecutive data points, and then 'moves' that window of data points forward over time. In business, moving averages are invaluable for understanding underlying sales trends, stock price movements, or website traffic patterns by reducing day-to-day 'noise'. They are often used for **forecasting**, where a future value is predicted based on the average of recent past values.

To evaluate the accuracy of forecasting models, including those using moving averages, the **Mean Absolute Deviation (MAD)** is a commonly used metric. MAD measures the average magnitude of the errors in a set of forecasts, without considering their direction. It is calculated as the average of the absolute differences between actual and forecasted values. A lower MAD indicates a more accurate forecast. While not directly a standard deviation, MAD is conceptually related to the **Standard Deviation (SD) of the forecast error**. For a normally distributed forecast error, the SD is approximately 1.25 times the MAD ($\sigma \approx 1.25 \times \text{MAD}$). This relationship allows forecasters to estimate the SD of errors from MAD, which can then be used to construct prediction intervals and assess forecast reliability.

### Moving Average (MA) Forecast Formula

The moving average forecast for a time series $Y$ at time $t$, with a window size of $n$, is calculated as the average of the $n$ most recent observations:

$\text{MA}_t = \frac{1}{n} \sum_{i=1}^{n} Y_{t-i}$

Where:
*   $\text{MA}_t$ is the moving average forecast for the current period.
*   $Y_{t-i}$ represents the actual value of the time series at past period $t-i$.
*   $n$ is the window size, i.e., the number of past observations included in the average.

### Mean Absolute Deviation (MAD) Formula

The Mean Absolute Deviation (MAD) measures the average magnitude of forecast errors, without considering their direction. It is calculated as:

$\text{MAD} = \frac{1}{N} \sum_{t=1}^{N} |Y_t - \hat{Y}_t|$

Where:
*   $N$ is the total number of forecasts.
*   $Y_t$ is the actual value at time $t$.
*   $\hat{Y}_t$ is the forecasted value at time $t$.
*   $|Y_t - \hat{Y}_t|$ is the absolute error between the actual and forecasted value at time $t$.

#### Understanding Rolling Window Options

When calculating a moving average (MA) with Pandas' `.rolling()` method, the `center` parameter dictates how the window is aligned:

*   **`center=True`**: The window is centered on the current data point. For an odd-sized window (e.g., `window=3`), the average for a specific year (e.g., 2015) will include the values from the previous year (2014), the current year (2015), and the next year (2016).

*   **`center=False` (default)**: The window is aligned to the right, meaning the current data point is the last element in the window. For a `window=3`, the MA for 2015 would be the average of 2013, 2014, and 2015. This is useful when you want to compute a statistic based only on past and current observations.

*   **Lagged Moving Average**: If you want the MA(3) for 2015 to represent the average of *only* previous years (e.g., 2012, 2013, 2014), you would typically combine the `.rolling()` method with a `.shift()` operation. First, calculate the rolling average with `center=False`, and then use `.shift(n)` (where `n` is the number of periods to shift) to move the calculated average to the desired past year. For example, `df['value'].rolling(window=3, center=False).mean().shift(1)` would give you an MA for 2015 that includes 2012, 2013, and 2014 (by shifting the 2014 average, which originally covered 2012-2014, to 2015).

Summary example:
| Method                           | Years used for MA at 2015 |
| -------------------------------- | ------------------------- |
| `rolling(3).mean()`              | 2013, 2014, 2015          |
| `rolling(3).mean().shift(1)`     | 2012, 2013, 2014          |
| `rolling(3, center=True).mean()` | 2014, 2015, 2016          |


````python
import matplotlib.pyplot as plt
import seaborn as sns
# Try time series analysis for loudness of pop songs over time
plt.figure(figsize=(12, 7))

# Calculate moving average of loudness with n=3
# Using 'loudness' column and shifting by 1 to get a lagged moving average forecast
df_pop['loudness_ma3'] = df_pop['loudness'].rolling(window=3).mean().shift(1)

# Plot original loudness
sns.lineplot(data=df_pop, x='year', y='loudness', errorbar=None, label='Actual Loudness')

# Plot 3-period moving average of loudness
sns.lineplot(data=df_pop, x='year', y='loudness_ma3', color='red', linestyle='--', label='3-Year Moving Average Forecast', errorbar=None)

plt.title('Loudness of Pop Songs Over Time with 3-Year Moving Average Forecast')
plt.xlabel('Year')
plt.ylabel('Loudness')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
````

````python
# calculate the MAD for the 3-period MA developed in the previous code block

# To calculate Mean Absolute Deviation (MAD) of the forecast error
# MAD = mean(|Actual - Forecast|)
# Here, 'loudness' is the actual value and 'loudness_ma3' is the forecast
mean_absolute_deviation = (df_pop['loudness'] - df_pop['loudness_ma3']).abs().mean()
print(f"Mean Absolute Deviation (MAD) of the 3-Year Moving Average forecast for loudness: {mean_absolute_deviation}")
````

````python
# Try out MA for loundness for window size from 3 to 30, report the MAD for each window size
for window_size in range(3, 31, 3):
  df_pop[f'loudness_ma{window_size}'] = df_pop['loudness'].rolling(window=window_size, min_periods=window_size).mean().shift(1)
  # Use a temporary variable for MAD calculation within the loop
  current_mad = (df_pop['loudness'] - df_pop[f'loudness_ma{window_size}']).abs().mean()
  print(f"Mean Absolute Deviation (MAD) of the {window_size}-Year Moving Average forecast for loudness: {current_mad}")
````

#### Understanding `min_periods` in `rolling()`

In pandas, when you apply the `.rolling()` method, the `min_periods` parameter specifies the minimum number of observations required in a window to perform the calculation. If there are fewer observations than `min_periods` in a given window, the result for that window will be `NaN` (Not a Number).

**Why is `min_periods` important?**

1.  **Handling Initial `NaN` Values**: By default, `min_periods` is set to 1. This means that a rolling calculation will be performed as soon as there's at least one data point in the window. However, this can sometimes lead to misleading results at the beginning of a time series, as early averages might be based on very few data points and thus not truly representative of the full window size.

2.  **Ensuring Data Integrity**: Setting `min_periods` to a value greater than 1 (often equal to the `window` size itself, as seen in the code `window=window_size, min_periods=window_size`) ensures that the calculated statistic (like a moving average) is always based on the full intended number of data points. This provides a more consistent and reliable measure throughout the series, especially when comparing different window sizes.

    *   For example, if `window=3` and `min_periods=3`, the first two calculations would result in `NaN` because there aren't three preceding data points to form a complete window. The calculation would only start from the third data point, using the first three values.

3.  **Reflecting True Average**: When performing a moving average forecast, as in the previous example, setting `min_periods` equal to the `window` ensures that each forecast truly represents an average of the specified number of past periods, avoiding forecasts based on incomplete information.

In the context of the previous moving average calculations for loudness:
```python
df_pop[f'loudness_ma{window_size}'] = df_pop['loudness'].rolling(window=window_size, min_periods=window_size).mean().shift(1)
```
Here, `min_periods=window_size` means that the moving average for `loudness` will only be calculated and appear (not as `NaN`) when there are at least `window_size` data points available in the rolling window. This ensures that every computed moving average is a 'full' average, based on the complete set of observations specified by the window size.

### (7.2) Autocorrelation and Autoregression: High-Level Concepts

In time series analysis, **autocorrelation** refers to the correlation of a time series with its own past values. Essentially, it measures how much the current observation of a variable is linearly related to its previous observations. A high autocorrelation suggests that past values have a strong predictive power for future values.

**Autoregression** builds upon this concept. An **Autoregressive (AR) model** is a linear model that predicts future values of a time series based on a weighted sum of its own past values. The idea is that the variable's past behavior can explain its future behavior. For example, an AR(1) model predicts the current value using only the immediately preceding value, while an AR(p) model uses the past 'p' values. Identifying the degree of autocorrelation helps in determining the appropriate 'p' (order) for an AR model, allowing us to capture the inherent temporal dependencies in the data for forecasting and understanding.

### Mathematical Equation of an AR(p) Model

An Autoregressive model of order `p`, denoted as AR(p), can be mathematically represented as:

$Y_t = c + \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + \dots + \phi_p Y_{t-p} + \epsilon_t$

Where:
*   $Y_t$ is the value of the time series at time `t`.
*   $c$ is a constant term (intercept).
*   $\phi_1, \phi_2, \dots, \phi_p$ are the autoregressive coefficients, which quantify the impact of past values on the current value.
*   $Y_{t-1}, Y_{t-2}, \dots, Y_{t-p}$ are the past `p` values of the time series.
*   $\epsilon_t$ is the white noise error term at time `t`, representing random shocks or disturbances that are not explained by the past values of $Y_t$.

The order `p` indicates how many previous observations are used to predict the current value. For example, an AR(1) model uses only $Y_{t-1}$, while an AR(2) model uses $Y_{t-1}$ and $Y_{t-2}$.

### Autoregression (AR) Models and Partial Autocorrelation Function (PACF)

**Autoregression (AR) Models** are a class of linear models that predict future values based on past observations. An AR model of order `p`, denoted as AR(p), predicts the current value of a time series as a linear combination of its past `p` values. For example, an AR(1) model uses the previous value to predict the current one, while an AR(2) uses the two previous values. These models are fundamental in time series analysis for capturing inherent dependencies and trends within sequential data.

To determine the appropriate order `p` for an AR model, the **Partial Autocorrelation Function (PACF)** is a crucial diagnostic tool. The PACF measures the correlation between a time series and its lagged values, after controlling for the effects of all intermediate lags. In simpler terms, PACF helps identify the direct relationship between an observation and its past values, excluding indirect correlations. A significant spike in the PACF at a particular lag `k` (and then drops to near zero) suggests that the time series can be modeled by an AR(k) process.

In Python, the `statsmodels` library provides the **`AutoReg` model**, which is specifically designed for fitting autoregressive time series models. This model allows you to specify the number of lags (`p`) directly. Once fitted, the `AutoReg` model can be used for forecasting and to understand the underlying autoregressive structure of the time series.

### Plot PACF and select the p.

````python
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_pacf
import pandas as pd

# Ensure yearly_loudness_mean is grouped by year
yearly_loudness_mean = df_pop.groupby('year')['loudness'].mean()

# Convert integer year index to PeriodIndex with yearly frequency, then to DatetimeIndex
# period index tells Python we have the same gap (getting the time& series ready before fitting the model)
yearly_loudness_mean.index = pd.PeriodIndex(yearly_loudness_mean.index, freq='Y')

# add timestamp on each period, so you have date time index
yearly_loudness_mean.index = yearly_loudness_mean.index.to_timestamp()

# Generate PACF plot
plt.figure(figsize=(8, 5))
# Limit lags to a safe maximum: min(20, (sample_size // 2) - 1)
# plot.pacf (要plot 的數值, lag, )
plot_pacf(yearly_loudness_mean, lags=min(20, (len(yearly_loudness_mean) // 2) - 1), ax=plt.gca())
plt.title('Partial Autocorrelation Function (PACF) for Yearly Mean Loudness')
plt.xlabel('Lags')
plt.ylabel('Partial Autocorrelation')
plt.grid(True)
plt.show()
````

The code snippets you provided are crucial for preparing time series data for analysis and then visualizing its autocorrelation properties:

*   **`yearly_loudness_mean_indexed.index = pd.PeriodIndex(yearly_loudness_mean_indexed.index, freq='Y')`**:
    *   This line converts the existing index of the `yearly_loudness_mean_indexed`, which is likely a simple integer representing the year, into a `PeriodIndex` with a yearly frequency (`freq='Y'`).
    *   A `PeriodIndex` represents time intervals (like years, months, etc.) rather than specific points in time. This is often useful for time series operations that deal with periods.
    
*   **`yearly_loudness_mean_indexed.index = yearly_loudness_mean_indexed.index.to_timestamp()`**:
    *   After creating a `PeriodIndex`, this line converts it to a `DatetimeIndex`.
    *   `DatetimeIndex` represents specific points in time. Many time series analysis functions, especially within libraries like `statsmodels`, prefer or require a `DatetimeIndex` to properly interpret the time dimension of the data.
    *   The combination of `PeriodIndex` and `to_timestamp()` ensures that the time series data has a clear, unambiguous time-based index that is compatible with advanced time series models.

*   **`plot_pacf(yearly_loudness_mean_indexed['loudness'], lags=min(20, (len(yearly_loudness_mean_indexed) // 2) - 1), ax=plt.gca())`**:
    *   This is the core line for generating the **Partial Autocorrelation Function (PACF)** plot.
    *   **`plot_pacf()`**: This function from `statsmodels.graphics.tsaplots` is used to visualize the PACF, which helps in identifying the order of an Autoregressive (AR) model.
    *   **`yearly_loudness_mean_indexed['loudness']`**: This specifies the time series data for which the PACF is to be calculated and plotted (in this case, the mean loudness values).
    *   **`lags=min(20, (len(yearly_loudness_mean_indexed) // 2) - 1)`**: This parameter determines the maximum number of lags to display on the plot. The `min()` function is used to set a reasonable upper bound for the lags:
        *   `20`: A common heuristic for the maximum number of lags to consider.
        *   `(len(yearly_loudness_mean_indexed) // 2) - 1`: This calculates half the length of the time series minus one. Plotting more than `N/2` lags (where N is the number of observations) is generally not recommended as the estimates become unreliable.
    *   **`ax=plt.gca()`**: This specifies that the plot should be drawn on the current Axes object, ensuring it integrates with any existing `matplotlib.pyplot` figure.

### Explanation of the AR(2) Model Summary

The `AutoReg` model summary provides a comprehensive overview of the fitted Autoregressive model. Let's break down the key sections:

**1. Model Information:**
*   **`Dep. Variable: loudness`**: Indicates that 'loudness' is the dependent variable being modeled.
*   **`Model: AutoReg(2)`**: Confirms that this is an Autoregressive model of order 2 (AR(2)), meaning the current value is predicted based on the two previous values.
*   **`Method: Conditional MLE`**: The method used for estimating the model parameters (Conditional Maximum Likelihood Estimation).
*   **`No. Observations: 50`**: The number of data points used in fitting the model.
*   **`S.D. of innovations: 1.202`**: The estimated standard deviation of the error term ($\epsilon_t$), also known as the residual standard error. A lower value indicates a better fit.
*   **`Log Likelihood: -76.926`**: A measure of the model's goodness of fit. Higher values indicate a better fit.
*   **`AIC, BIC, HQIC`**: Information criteria (Akaike, Bayesian, Hannan-Quinn) used for model selection. Lower values generally indicate a better model, balancing fit and complexity.

**2. Coefficients Table:**
This table shows the estimated parameters of the AR(2) model:
*   **`const`**: This is the intercept term. Its value is -1.8570. The **P>|z|** (p-value) for the constant is 0.043, which is less than 0.05, suggesting that the intercept is statistically significant at the 5% level.
*   **`loudness.L1`**: This is the coefficient for the first lag ($Y_{t-1}$). Its value is 0.4034. The **P>|z|** is 0.002, which is highly significant, indicating that the mean loudness from the previous year has a strong positive influence on the current year's mean loudness.
*   **`loudness.L2`**: This is the coefficient for the second lag ($Y_{t-2}$). Its value is 0.3489. The **P>|z|** is 0.008, which is also highly significant, indicating that the mean loudness from two years ago also has a strong positive influence on the current year's mean loudness.

**Interpretation of Coefficients:**
The positive and significant coefficients for `loudness.L1` and `loudness.L2` suggest that there is a persistent trend in the mean loudness of pop songs. Specifically, if the loudness was higher in the past one or two years, it tends to remain higher in the current year, and vice-versa. The current mean loudness is an autoregressive process where roughly 40% of the previous year's loudness and 35% of the loudness from two years ago contribute to the current year's loudness, after accounting for the constant term.

**3. Roots:**

If the imaginary part of an AR root is non-zero, it indicates that there is oscillatory or cyclical behavior in the time series. ### 1️⃣ **SD of Innovation in an AR Model**

Consider an AR(p) model:

$$Y_t = \phi_1 Y_{t-1} + \dots + \phi_p Y_{t-p} + \varepsilon_t$$

where
$\varepsilon_t$ = **innovation** (or shock) at time $t$

### What is the innovation?

The innovation is:

*   The new information arriving at time $t$
*   The unpredictable component of $Y_t$
*   A white noise process:
    $\varepsilon_t \sim (0, \sigma^2)$

The **standard deviation of innovation** is:

$$\sigma = \sqrt{\text{Var}(\varepsilon_t)}$$

It represents:

*   The inherent randomness of the process
*   The size of unpredictable shocks
*   The theoretical minimum one-step-ahead forecast error (if parameters are known)

In other words:

> It measures the noise level in the time series.

Even with a perfectly specified AR model, you cannot forecast better than this noise level.

---

## 2️⃣ **RMSE (Root Mean Squared Error)**

RMSE is an empirical measure of forecast accuracy:

$RMSE = \sqrt{\frac{1}{n} \sum_{t=1}^n (Y_t - \hat{Y}_t)^2}$

It measures:

*   How far predictions are from actual values
*   Overall prediction accuracy
*   Both model misspecification and random noise

---

## 3️⃣ **Difference Between SD of Innovation and RMSE**

| SD of Innovation                           | RMSE                          |
| :----------------------------------------- | :---------------------------- |
| Theoretical quantity                       | Empirical quantity            |
| Measures true noise level                  | Measures forecast performance |
| Comes from model error term                | Comes from actual predictions |
| Minimum achievable one-step forecast error | Usually $\geq$ SD of innovation    |

### Key Relationship

For a correctly specified AR model:

$$\text{One-step forecast RMSE} \approx \sigma$$

If:

*   RMSE $\approx$ SD of innovation $\rightarrow$ model is well specified
*   RMSE $>$ SD of innovation $\rightarrow$ model is missing structure
*   RMSE $<$ SD of innovation $\rightarrow$ something is wrong (data leakage or calculation issue)

---

## 4️⃣ Intuition

Think of it this way:

*   **SD of innovation** = unavoidable randomness
*   **RMSE** = actual forecasting performance
*   The difference tells you how much error comes from model imperfection vs. pure randomness

#### More about stationarity

#### **Consistent Statistical Behavior:** The core statistical characteristics of the loudness series – its average value (mean), how much it varies (variance), and how past loudness values influence current ones (autocovariance) – remain constant over time. This means these properties don't change or drift in a systematic way.

#### **No Significant Underlying Trends or Varying Volatility:** It tells us that there isn't an unmodeled, long-term upward or downward trend in loudness, nor is the variability of loudness consistently increasing or decreasing over the years. The model itself, with its autoregressive components, captures the dependencies adequately without needing to account for non-stationary factors.

#### **Reliable Historical Patterns for Prediction:** Because the series is stationary, the relationships observed between a year's loudness and its two preceding years' loudness (as quantified by the model's coefficients) are stable. This stability is crucial because it allows us to trust that these patterns from the past can be reliably used to understand and potentially forecast future loudness levels without the need for additional transformations.

####In essence, finding that the loudness time series is stationary provides confidence that the AR(2) model is robust and that its findings accurately reflect the underlying, consistent dynamics of pop song loudness over time.

#### Assumptions of the Autoregressive (AR) Model

For an Autoregressive (AR) model to be valid and produce reliable results, several key assumptions about the underlying time series and its error term must be met:

1.  **Stationarity**: As discussed, the time series must be stationary. This means its mean, variance, and autocovariance structure should remain constant over time. If the series is non-stationary, appropriate transformations (like differencing) are usually required before fitting an AR model.

2.  **Linearity**: The current value of the time series is assumed to be a linear function of its past values. This means the relationships between $Y_t$ and $Y_{t-k}$ are linear.

3.  **Independence of Error Terms (White Noise)**: The error terms (residuals) $\epsilon_t$ must be independent and identically distributed (i.i.d.) with a mean of zero and constant variance (white noise). This implies that there is no remaining pattern or autocorrelation in the residuals that the AR model failed to capture. We often check this assumption by examining the autocorrelation function (ACF) of the residuals.

4.  **Homoscedasticity**: The variance of the error terms ($\epsilon_t$) should be constant across all observations. This is often referred to as homoscedasticity.

5.  **Normality of Error Terms (Optional but Desirable)**: While not strictly required for the model to function, it is often assumed that the error terms are normally distributed. This assumption is particularly important for constructing confidence intervals and performing hypothesis tests on the model coefficients.

Violations of these assumptions can lead to biased coefficient estimates, inefficient forecasts, and incorrect statistical inferences. Therefore, it's crucial to assess these assumptions through diagnostic checks after fitting an AR model.
